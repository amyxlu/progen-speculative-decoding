```python
ppl_by_len = {

}
for k in tqdm(res):
    s = res[k]
    ppls = []

    for i in range(1, len(s)+1):
        ppls.append(RITA_perplexity.calc_perplexity(s[:i]))
    
    ppl_by_len[k] = ppls
```

```python
import matplotlib.pyplot as plt

for k in ppl_by_len:
    plt.plot(ppl_by_len[k], label=f"# Draft tokens={k}")

plt.legend()
```
